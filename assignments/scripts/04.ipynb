```python
# 运算符
# 算术运算符
x=5
y=3
print(x+y) # 加法
print(x-y) # 减法  
print(x*y) # 乘法
print(x/y) # 除法
print(x//y) # 整除
print(x%y) # 取余
print(x**y) # 幂运算

# 比较运算符
print(x==y) # 等于
print(x!=y) # 不等于
print(x>y) # 大于
print(x<y) # 小于
print(x>=y) # 大于等于
print(x<=y) # 小于等于

# 赋值运算符
z=x+y #等价于 z=x+y
z+=x  #等价于 z=z+x
z-=x  #等价于 z=z-x
z*=x  #等价于 z=z*x
z/=x  #等价于 z=z/x
z%=x  #等价于 z=z%x
z//=x #等价于 z=z//x
z**=x #等价于 z=z**x

# 逻辑运算符
print(x>0 and y>0)#与
print(x>0 or y>0) #或
print(not(x>0))   #非

# 成员运算符
print(x in [1,2,3,4,5]) #判断x是否在列表中
print(x not in [1,2,3,4,5]) #判断x是否不在列表中
```

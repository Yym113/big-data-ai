```python
# 运算符

# 算术运算符
n1=10
n2=3
print(n1+n2,n1-n2,n1*n2,n1/n2,n1//n2,n1%n2,n1**n2)

# 赋值运算符
n3=5
n3+=n3
print(n3)

# 比较运算符
n4=10
print(n4>=10,n4<15,n4>5,n4<=5,n4==10,n4!=5,n4<=20)

# 逻辑运算符
n5=True
print(n5 and False,n5 or False,not n5)

# 成员运算符
Iist1=[1,2,3,4,5]
print(3 in Iist1,6 not in Iist1)
```

```python
# 字符串的格式化
name="小明"
age=19

# 使用format方法
print("大家好,我是{0}，今年{1}岁。".format(name,age))

# 使用f-string格式化字符串
print(f"我的名字是{name}，我今年{age}岁。")
```

```python
## 分支

age=16
if age>=18:
    print("你已成年了")
    print("你已经可以独立生活了")
elif age>=13:
    print("你是青少年") 
else:
    print("你还未成年")
```

```python
while True:
    user_input=input("请输入一个数字：(输入'exit'退出):")
    if user_input=="exit":
        print("你输入的数字是：",user_input)
        break
    try:
        number=int(user_input)
        print(f"你输入的数字是：{number}")
    except ValueError:
        print("输入无效，请重新输入一个数字。")
```

```python
hunger_level = 5
while hunger_level > 0:
    print("我饿了，开始吃东西")
    hunger_level -= 1
    print(f"当前饥饿程度：{hunger_level}")
```
